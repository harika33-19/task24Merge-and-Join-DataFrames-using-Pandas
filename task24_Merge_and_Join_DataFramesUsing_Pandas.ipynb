```python
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4],
    "Customer_Name": ["Asha", "Rahul", "Priya", "Kiran"],
    "City": ["Hyderabad", "Guntur", "Vijayawada", "Chennai"]
})

print("Customer DataFrame:")
print(customers)
```

```python
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Customer_ID": [1, 2, 2, 5],
    "Amount": [1500, 2500, 1800, 3000]
})

print("Orders DataFrame:")
print(orders)
```

```python
inner_join = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="inner"
)

print("INNER JOIN RESULT:")
print(inner_join)
```

```python
left_join = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)

print("LEFT JOIN RESULT:")
print(left_join)
```

```python
right_join = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="right"
)

print("RIGHT JOIN RESULT:")
print(right_join)
```

```python
outer_join = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer"
)

print("OUTER JOIN RESULT:")
print(outer_join)
```

```python
comparison = pd.DataFrame({
    "Join_Type": ["Inner Join", "Left Join", "Right Join", "Outer Join"],
    "Number_of_Rows": [
        len(inner_join),
        len(left_join),
        len(right_join),
        len(outer_join)
    ],
    "Description": [
        "Only matching records",
        "All customers and matching orders",
        "All orders and matching customers",
        "All customers and all orders"
    ]
})

print(comparison)
```

```python
merge_status = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer",
    indicator=True
)

print(merge_status)
```

```python
unmatched_records = merge_status[
    merge_status["_merge"] != "both"
]

print("UNMATCHED RECORDS:")
print(unmatched_records)
```

```python
final_dataset = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer"
)

print("FINAL COMBINED DATASET:")
print(final_dataset)
```

```python
final_dataset.to_csv(
    "final_combined_dataset.csv",
    index=False
)

print("Final dataset saved successfully!")
```

```python
import pandas as pd

# 1. Create customer DataFrame
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4],
    "Customer_Name": ["Asha", "Rahul", "Priya", "Kiran"],
    "City": ["Hyderabad", "Guntur", "Vijayawada", "Chennai"]
})

# 2. Create orders DataFrame
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Customer_ID": [1, 2, 2, 5],
    "Amount": [1500, 2500, 1800, 3000]
})

# 3. Display original DataFrames
print("CUSTOMERS:")
print(customers)

print("\nORDERS:")
print(orders)

# 4. Inner Join
inner_join = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="inner"
)

print("\nINNER JOIN:")
print(inner_join)

# 5. Left Join
left_join = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="left"
)

print("\nLEFT JOIN:")
print(left_join)

# 6. Right Join
right_join = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="right"
)

print("\nRIGHT JOIN:")
print(right_join)

# 7. Outer Join
outer_join = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="outer"
)

print("\nOUTER JOIN:")
print(outer_join)

# 8. Compare join results
comparison = pd.DataFrame({
    "Join_Type": [
        "Inner Join",
        "Left Join",
        "Right Join",
        "Outer Join"
    ],
    "Number_of_Rows": [
        len(inner_join),
        len(left_join),
        len(right_join),
        len(outer_join)
    ]
})

print("\nJOIN COMPARISON:")
print(comparison)

# 9. Identify unmatched records
merge_status = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer",
    indicator=True
)

unmatched_records = merge_status[
    merge_status["_merge"] != "both"
]

print("\nUNMATCHED RECORDS:")
print(unmatched_records)

# 10. Save original datasets
customers.to_csv("customers.csv", index=False)
orders.to_csv("orders.csv", index=False)

# 11. Save final combined dataset
outer_join.to_csv(
    "final_combined_dataset.csv",
    index=False
)

print("\nAll datasets saved successfully!")
```
